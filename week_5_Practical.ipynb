```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
# Upload CSV file
from google.colab import files
uploaded = files.upload()

# Get the uploaded file name
CSV_PATH = list(uploaded.keys())[0]

# Read CSV file
df = pd.read_csv(CSV_PATH)
# Select only two numerical features
X = df[['CodingTestScore', 'AptitudeTestScore']]

# Apply Standard Scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=['CodingTestScore', 'AptitudeTestScore'])

print("Original Data")
print(X.head())

print("\nAfter standard Scaling")
print(scaled_df.head())
```

```python
import pandas as pd
from google.colab import files
# Upload CSV file
uploaded = files.upload()

# Get the uploaded file name
CSV_PATH = list(uploaded.keys())[0]

# Read CSV file
df = pd.read_csv(CSV_PATH)
# Fill missing categorical values with mode
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0]) # Changed to reassign instead of inplace=True

df.to_csv("placementfilledcat.csv", index=False)
print(df.isnull().sum())
```
